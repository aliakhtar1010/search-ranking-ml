## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## Decision
Help content editors decide which webpages should be reviewed first for refresh.

## Action
Content editors or SEO managers will use the ranked recommendations to decide whether a webpage should be refreshed, expanded, monitored, or left unchanged.

## Cost of a Wrong Recommendation
A false positive may waste editorial time by recommending a healthy page for review. A false negative may cause a declining page to be overlooked, leading to continued traffic loss and missed opportunities to improve search performance.

## Why Machine Learning?
A simple rule can identify some pages, but webpage performance depends on multiple interacting signals such as impressions, CTR, content age, and engagement. Machine learning can combine these signals to produce a more effective ranking while still being evaluated against a simple baseline.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains **30,000 webpages**, and approximately **54.2% are labelled as declining**. This shows that prioritizing pages for review is a meaningful problem. The correlation between search volume and impressions is approximately **0.001**, suggesting that search volume alone does not explain page visibility well. This supports using multiple search, content, and engagement signals when ranking pages for review.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import subprocess
import pandas as pd

# Clone your repository into this fresh Colab session
REPO_URL = "https://github.com/aliakhtar1010/search-ranking-ml.git"
REPO_DIR = "/content/search-ranking-ml"

if not os.path.exists(REPO_DIR):
    subprocess.run(["git", "clone", REPO_URL, REPO_DIR], check=True)

os.chdir(REPO_DIR)

# Load the starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Show 2–3 real numbers that support the lane choice
total_pages = len(df)
declining_pages = (df["trend_direction"].str.lower() == "down").sum()
declining_rate = (df["trend_direction"].str.lower() == "down").mean() * 100
correlation = df["search_volume"].corr(df["impressions_90d"])

print(f"Total pages: {total_pages}")
print(f"Declining pages: {declining_pages}")
print(f"Declining rate: {declining_rate:.1f}%")
print(f"Search volume vs impressions correlation: {correlation:.3f}")
```
